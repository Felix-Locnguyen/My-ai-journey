2. Regularized Regression (Ridge, Lasso & Elastic Net)
- Lý thuyết & Nguyên lý: Thêm thành phần phạt vào hàm mất mát MSE để giảm hiện tượng quá khớp (overfitting) và xử lý đa cộng tuyến 

- Ridge ($L_2$): Phạt bình phương độ lớn trọng số $\lambda \sum \beta_j^2$, thu nhỏ trọng số liên tục về gần.

- Lasso ($L_1$): Phạt giá trị tuyệt đối $\lambda \sum |\beta_j|$, có khả năng triệt tiêu trọng số về đúng 0 để tự động chọn lọc đặc trưng. 

- Elastic Net: Kết hợp cả $L_1$ và $L_2$

```python
import matplotlib.pyplot as plt
import numpy as np

# Khai báo dải giá trị alpha (lambda) theo thang logarit
alphas = np.logspace(-2, 2, 50)

# Minh họa độ lớn trọng số khi alpha/lambda tăng
plt.figure(figsize=(7, 4))

# Ridge (L2): Thu nhỏ mượt mà trọng số nhưng không triệt tiêu hẳn về 0
plt.plot(alphas, 1 / (1 + alphas), label='Ridge (L2 - Thu nhỏ mượt)')

# Lasso (L1): Triệt tiêu trọng số về 0 khi alpha tăng cao (dùng np.clip hoặc np.maximum để giới hạn đoạn >= 0)
plt.plot(
    alphas,
    np.maximum(0, 1 - np.log10(alphas + 1)),
    label='Lasso (L1 - Triệt tiêu về 0)',
)

plt.xscale('log')
plt.title('Sự thay đổi trọng số theo hệ số phạt Lambda (Alpha)')
plt.xlabel('Hệ số phạt Lambda')
plt.ylabel('Hệ số trọng số (Coefficients)')
plt.legend()
plt.grid(True)
plt.show()
```

Ý nghĩa của biểu đồ này:
- Ridge (L2 Regularization): Đường màu xanh thể hiện trọng số giảm dần một cách mượt mà và tiến dần về gần 0 khi hệ số phạt $\lambda$ tăng lên, nhưng hiếm khi chạm hẳn về bằng 0.

- Lasso (L1 Regularization): Đường màu cam thể hiện khả năng đặc biệt của Lasso là có thể triệt tiêu hoàn toàn các trọng số về 0, giúp tự động loại bỏ các đặc trưng không quan trọng (feature selection).
